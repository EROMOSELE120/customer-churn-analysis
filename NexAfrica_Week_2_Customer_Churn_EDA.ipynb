```python
import pandas as pd

df = pd.read_csv('/content/cleaned_customer_churn.csv')

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
import matplotlib.pyplot as plt

numeric_columns = ['tenure', 'MonthlyCharges', 'TotalCharges']

df[numeric_columns].boxplot(figsize=(10, 6))

plt.title('Boxplots of Numerical Features')
plt.ylabel('Values')
plt.show()
```

```python
numeric_columns = ['tenure', 'MonthlyCharges', 'TotalCharges']

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"\n{column}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

No potential outliers were detected in the numerical variables (tenure, MonthlyCharges, and TotalCharges) using the IQR method. Therefore, no outlier removal was necessary.

```python
categorical_columns = df.select_dtypes(include='object').columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Convert the target variable Churn into numbers
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})

print(df['Churn'].value_counts())
```

```python
# Get the categorical columns, excluding Churn
categorical_features = df.select_dtypes(include='object').columns

# Convert categorical columns into 0/1 columns
df_encoded = pd.get_dummies(
    df,
    columns=categorical_features,
    drop_first=True
)

print("New shape:", df_encoded.shape)

print("\nFirst 5 rows:")
display(df_encoded.head())
```

```python
print("Missing values:", df_encoded.isnull().sum().sum())
```

```python
df_encoded['TenureGroup'] = pd.cut(
    df_encoded['tenure'],
    bins=[-1, 12, 24, 48, 72],
    labels=['0-12 Months', '13-24 Months', '25-48 Months', '49-72 Months']
)

print(df_encoded['TenureGroup'].value_counts())
```

```python
df_encoded['TenureGroup'] = df_encoded['TenureGroup'].map({
    '0-12 Months': 1,
    '13-24 Months': 2,
    '25-48 Months': 3,
    '49-72 Months': 4
})

print(df_encoded['TenureGroup'].value_counts().sort_index())
```

```python
print(df_encoded.columns.tolist())
```

```python
X = df_encoded.drop('Churn', axis=1)
y = df_encoded['Churn']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(X.dtypes.value_counts())
```

```python
X = X.copy()

# Convert True/False columns to 0/1
X = X.astype(int)

print(X.dtypes.value_counts())
```

```python
final_dataset = pd.concat([X, y], axis=1)

final_dataset.to_csv('/content/churn_ml_ready.csv', index=False)

print("Final dataset shape:", final_dataset.shape)
print("File saved successfully!")
```

## Key Data-Driven Insights

## Feature Engineering

The categorical variables were converted into numerical variables using one-hot encoding so that they could be used by machine-learning algorithms.

The target variable, Churn, was encoded as:
- 0 = No churn
- 1 = Churn

A new feature called TenureGroup was created from the tenure variable. Customers were grouped into:
- 0–12 Months
- 13–24 Months
- 25–48 Months
- 49–72 Months

The final feature dataset contains 31 numerical features, while Churn is retained as the target variable.

## Conclusion

The Exploratory Data Analysis revealed several important patterns in customer churn. Contract type, tenure, and monthly charges showed notable relationships with churn. Month-to-month customers had a substantially higher churn rate than customers on longer-term contracts, while customers who churned generally had shorter tenure and higher monthly charges.

Potential outliers were checked using the IQR method, and no outliers were detected in tenure, MonthlyCharges, or TotalCharges.

Categorical variables were encoded, a TenureGroup feature was created, and the features were converted into numerical format. The resulting dataset is prepared for use in machine-learning models.
